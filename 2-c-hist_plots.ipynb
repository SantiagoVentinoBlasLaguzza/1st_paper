```python
import os
import yaml
import torch
import pandas as pd
import numpy as np
import random
import logging
import matplotlib.pyplot as plt

from sklearn.model_selection import StratifiedKFold, train_test_split

logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")

# ------------------------------------------------------------------------
# CONFIGURACIÓN Y RUTAS
# ------------------------------------------------------------------------
project_dir = '/home/diego/Escritorio/santiago/1st_paper/116ROIs'
csv_path = os.path.join(project_dir, 'DataBaseSubjects.csv')
tensor_data_dir = os.path.join(project_dir, 'TensorData')

# Cargar config (seed, etc.)
config_path = os.path.join(project_dir, 'config.yaml')
with open(config_path, 'r') as f:
    config = yaml.safe_load(f)

seed = config.get('seed', 42)
np.random.seed(seed)
torch.manual_seed(seed)
random.seed(seed)

# Leer CSV
subjects_df = pd.read_csv(csv_path)

# (Opcional) unificar MCI, EMCI, LMCI en “Others” si lo deseas
def map_3class_group(x):
    if x == 'AD':
        return 'AD'
    elif x == 'CN':
        return 'CN'
    else:
        return 'Others'
# subjects_df['ResearchGroup'] = subjects_df['ResearchGroup'].apply(map_3class_group)

# Crear etiqueta combinada (ResearchGroup + Sex)
subjects_df['Group_Sex'] = subjects_df['ResearchGroup'].astype(str) + '_' + subjects_df['Sex'].astype(str)

# Construir un diccionario con rutas a .pt
grouped_data = subjects_df.groupby(['ResearchGroup','Sex'])['SubjectID'].apply(list).to_dict()

tensor_groups = {}
for (group, sex), subject_ids in grouped_data.items():
    file_paths = []
    for sid in subject_ids:
        if not os.path.exists(tensor_data_dir):
            logging.warning(f"Carpeta inexistente: {tensor_data_dir}")
            continue
        if group not in ['AD', 'CN']:
            group = 'Others'
        fp = os.path.join(tensor_data_dir, f"{group}_tensor_{sid}.pt")
        file_paths.append(fp)
    tensor_groups[f"{group}_{sex}"] = file_paths

# ------------------------------------------------------------------------
# FUNCIONES DE PREPROCESADO
# ------------------------------------------------------------------------

def zero_diagonals(tensor: torch.Tensor) -> torch.Tensor:
    """
    Pone a 0 todas las diagonales de cada canal.
    """
    C, H, W = tensor.shape
    diag_idx = torch.arange(H)
    for c in range(C):
        tensor[c, diag_idx, diag_idx] = 0.0
    return tensor

def apply_channel_transforms(tensor: torch.Tensor) -> torch.Tensor:
    """
    Ejemplo: fisher-z a canal 0 (Correlación), log(1+x) a canal 2 (Granger).
    Ajusta según tus necesidades.
    """
    eps = 1e-8
    # Canal 0 => Fisher Z
    corr_values = torch.clamp(tensor[0], -0.9999, 0.9999)
    fisher = 0.5 * torch.log((1 + corr_values)/(1 - corr_values + eps))
    tensor[0] = fisher

    # Canal 2 => log(1 + x) para Granger
    gc_nonneg = torch.clamp(tensor[2], min=0.0)
    gc_log = torch.log(gc_nonneg + 1.0)
    tensor[2] = gc_log

    return tensor

def load_tensor(fp):
    """
    Carga un tensor .pt, con shape (4,116,116).
    Aplica zero_diagonals y apply_channel_transforms.
    """
    if not os.path.exists(fp):
        logging.warning(f"No existe: {fp}")
        return None
    try:
        data = torch.load(fp)  # shape (4,116,116) (o lo que sea)
        if isinstance(data, np.ndarray):
            data = torch.tensor(data, dtype=torch.float32)
        if not isinstance(data, torch.Tensor):
            logging.warning(f"Formato inesperado en {fp}: {type(data)}")
            return None

        data = zero_diagonals(data)
        data = apply_channel_transforms(data)
        return data
    except Exception as e:
        logging.error(f"Error al cargar {fp}: {e}")
        return None

def load_tensors(paths, batch_size=16):
    """
    Carga en batches para evitar problemas de RAM en grandes datasets.
    Retorna (N,4,116,116) o None.
    """
    all_tensors = []
    for i in range(0, len(paths), batch_size):
        batch_paths = paths[i:i+batch_size]
        batch_data = []
        for p in batch_paths:
            t = load_tensor(p)
            if t is not None:
                batch_data.append(t)
        if batch_data:
            all_tensors.extend(batch_data)
    if len(all_tensors) == 0:
        return None
    return torch.stack(all_tensors, dim=0)  # => (N,4,116,116)

def compute_mean_std_ignoring_diagonal(dataset: torch.Tensor):
    """
    dataset: (N,C,H,W). Ignora diagonales en cada canal
    para calcular mean y std. Retorna shape (C,).
    """
    N, C, H, W = dataset.shape
    diag_mask = torch.ones(H, W, dtype=bool)
    idx = torch.arange(H)
    diag_mask[idx, idx] = False  # la diagonal => False

    means = torch.zeros(C, dtype=torch.float32)
    stds  = torch.zeros(C, dtype=torch.float32)

    for c in range(C):
        channel_data = dataset[:, c, :, :]  # (N,H,W)
        offdiag = channel_data[:, diag_mask]  # (N * #offdiag)
        m = offdiag.mean()
        s = offdiag.std()
        if s < 1e-12:
            s = 1e-9
        means[c] = m
        stds[c]  = s
    return means, stds

def zscore_ignoring_diagonal(dataset: torch.Tensor, mean_c: torch.Tensor, std_c: torch.Tensor):
    """
    z-score canal por canal, ignorando la diagonal en el cálculo de mean/std
    (ya calculados), pero aquí la diagonal no se reescribe (permanece = 0).
    """
    out = dataset.clone()
    eps = 1e-9
    for c in range(dataset.shape[1]):
        out[:, c] = (out[:, c] - mean_c[c]) / (std_c[c] + eps)
    return out


# ------------------------------------------------------------------------
# FUNCIONES PARA HISTOGRAMAS IGNORANDO DIAGONAL
# ------------------------------------------------------------------------

def flatten_off_diagonal(data: torch.Tensor, c: int) -> np.ndarray:
    """
    Toma data shape=(N,C,H,W). Devuelve un vector con
    SOLO los valores off-diagonal del canal c en TODOS los sujetos.
    """
    N, C_, H, W = data.shape
    assert c < C_, "Canal fuera de rango"
    diag_mask = np.ones((H, W), dtype=bool)
    np.fill_diagonal(diag_mask, False)  # la diagonal => False
    channel_slice = data[:, c, :, :]  # => shape (N,H,W)
    off_diag_vals = channel_slice[:, diag_mask]  # => shape (N * #offdiag)
    return off_diag_vals.flatten().cpu().numpy()

def plot_histograms_by_channel(
    data_train_pre, data_val_pre, data_test_pre,
    data_train_post, data_val_post, data_test_post,
    fold_idx
):
    """
    4 canales => 2 filas (pre/post) x 4 columnas (cada canal).
    Hist. con 3 colores: train, val, test, ignorando diagonales.
    """
    channel_names = ["Correlation", "Mutual Info", "Granger Causality", "DCV"]
    C = data_train_pre.shape[1]

    fig, axes = plt.subplots(nrows=2, ncols=C, figsize=(5*C, 10))
    fig.suptitle(f"Histogramas (sin diagonales) - Fold {fold_idx}", fontsize=15)

    def plot_hist(ax, values, color, label, bins=50, alpha=0.35):
        ax.hist(values, bins=bins, alpha=alpha, color=color, label=label, density=True, edgecolor='black')

    for c in range(C):
        # Fila 0 => pre
        ax0 = axes[0, c]
        tr_pre = flatten_off_diagonal(data_train_pre, c)
        vl_pre = flatten_off_diagonal(data_val_pre,   c)
        ts_pre = flatten_off_diagonal(data_test_pre,  c)
        plot_hist(ax0, tr_pre, 'blue',   'Train')
        plot_hist(ax0, vl_pre, 'orange', 'Val')
        plot_hist(ax0, ts_pre, 'green',  'Test')
        ax0.set_title(f"{channel_names[c]} (Pre)")
        ax0.set_xlabel("Valor")
        ax0.set_ylabel("Densidad")
        ax0.legend()

        # Fila 1 => post
        ax1 = axes[1, c]
        tr_post = flatten_off_diagonal(data_train_post, c)
        vl_post = flatten_off_diagonal(data_val_post,   c)
        ts_post = flatten_off_diagonal(data_test_post,  c)
        plot_hist(ax1, tr_post, 'blue',   'Train')
        plot_hist(ax1, vl_post, 'orange', 'Val')
        plot_hist(ax1, ts_post, 'green',  'Test')
        ax1.set_title(f"{channel_names[c]} (Post)")
        ax1.set_xlabel("Valor")
        ax1.set_ylabel("Densidad")
        ax1.legend()

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

# ------------------------------------------------------------------------
# PLOT DE 4 MATRICES ANTES vs DESPUÉS PARA UN SUJETO RANDOM EN TRAIN
# ------------------------------------------------------------------------
def plot_four_matrices_before_after(
    matrix_before: torch.Tensor,
    matrix_after: torch.Tensor,
    fold_idx: int,
    subject_idx: int
):
    """
    Muestra 2 filas x 4 columnas:
      - fila 0 => 4 canales (before)
      - fila 1 => 4 canales (after)
    matrix_before, matrix_after => shape (4,116,116)
    """
    titles = ["Correlation", "Mutual Info", "Granger Causality", "DCV"]
    fig, axes = plt.subplots(nrows=2, ncols=4, figsize=(20, 8))
    fig.suptitle(f"Fold {fold_idx} - Sujeto train_idx={subject_idx}", fontsize=15)

    # Fila 0 => before
    for i in range(4):
        ax = axes[0, i]
        im = ax.imshow(matrix_before[i].cpu().numpy(), aspect='auto', cmap='jet')
        ax.set_title(f"{titles[i]} (Pre)")
        plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

    # Fila 1 => after
    for i in range(4):
        ax = axes[1, i]
        im = ax.imshow(matrix_after[i].cpu().numpy(), aspect='auto', cmap='jet')
        ax.set_title(f"{titles[i]} (Post)")
        plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

    plt.tight_layout(rect=[0, 0, 1, 0.94])
    plt.show()

# ------------------------------------------------------------------------
# CROSS-VALIDATION: 5 FOLDS
# ------------------------------------------------------------------------
# Recolectar rutas y etiquetas
X = []
y = []
for gsex, file_paths in tensor_groups.items():
    for fp in file_paths:
        X.append(fp)
        y.append(gsex)

X = np.array(X)
y = np.array(y)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
fold_number = 1

for train_val_idx, test_idx in skf.split(X, y):
    logging.info(f"\n===== FOLD {fold_number} =====")
    X_train_val, X_test = X[train_val_idx], X[test_idx]
    y_train_val, y_test = y[train_val_idx], y[test_idx]

    # Separar train vs val (80/20)
    train_idx, val_idx = train_test_split(
        np.arange(len(X_train_val)),
        test_size=0.2,
        stratify=y_train_val,
        random_state=seed
    )
    X_train, X_val = X_train_val[train_idx], X_train_val[val_idx]
    y_train, y_val = y_train_val[train_idx], y_train_val[val_idx]

    # Cargar tensores
    train_data = load_tensors(list(X_train))
    val_data   = load_tensors(list(X_val))
    test_data  = load_tensors(list(X_test))

    if train_data is None or val_data is None or test_data is None:
        logging.warning("Faltan datos en este fold. Se omite.")
        fold_number += 1
        continue

    # Copias pre-zscore
    train_pre = train_data.clone()
    val_pre   = val_data.clone()
    test_pre  = test_data.clone()

    # Mean/std (ignorando diagonal) SOLO en train
    mean_c, std_c = compute_mean_std_ignoring_diagonal(train_data)

    # Normalizar
    train_data = zscore_ignoring_diagonal(train_data, mean_c, std_c)
    val_data   = zscore_ignoring_diagonal(val_data,   mean_c, std_c)
    test_data  = zscore_ignoring_diagonal(test_data,  mean_c, std_c)

    # 1) Histogramas pre/post (train, val, test) ignorando diagonales
    plot_histograms_by_channel(
        data_train_pre=train_pre,
        data_val_pre=val_pre,
        data_test_pre=test_pre,
        data_train_post=train_data,
        data_val_post=val_data,
        data_test_post=test_data,
        fold_idx=fold_number
    )

    # 2) Elegir sujeto random del train y graficar sus 4 canales (pre vs post)
    if train_data.size(0) > 0:
        random_idx = random.randint(0, train_data.size(0)-1)
        subj_matrix_pre  = train_pre[random_idx]  # shape (4,116,116)
        subj_matrix_post = train_data[random_idx] # shape (4,116,116)
        plot_four_matrices_before_after(
            matrix_before=subj_matrix_pre,
            matrix_after=subj_matrix_post,
            fold_idx=fold_number,
            subject_idx=random_idx
        )

    # (Opcional) Guardar si quieres
   # fold_dir = os.path.join(project_dir, f"fold_{fold_number}")
   # os.makedirs(fold_dir, exist_ok=True)
   # torch.save(train_data, os.path.join(fold_dir, 'train_data_normed.pt'))
   # torch.save(val_data,   os.path.join(fold_dir, 'val_data_normed.pt'))
   # torch.save(test_data,  os.path.join(fold_dir, 'test_data_normed.pt'))
   # logging.info(f"Guardados tensores normalizados del fold {fold_number} en {fold_dir}")

    

        # ...
    # (Opcional) Guardar tensores y etiquetas
    fold_dir = os.path.join(project_dir, f"fold_{fold_number}")
    os.makedirs(fold_dir, exist_ok=True)

    # Guardar tensores normalizados
    torch.save(train_data, os.path.join(fold_dir, 'train_data_normed.pt'))
    print("train_data", train_data.shape)
    torch.save(val_data,   os.path.join(fold_dir, 'val_data_normed.pt'))
    print("val_data", val_data.shape)
    torch.save(test_data,  os.path.join(fold_dir, 'test_data_normed.pt'))
    print("test_data", test_data.shape)

    # Guardar etiquetas (pueden ser lista/array de strings)
    # y_train, y_val y y_test deben ser del mismo largo que train_data, val_data y test_data, respectivamente
    torch.save(y_train, os.path.join(fold_dir, f'train_labels_fold_{fold_number}.pt'))
    print("y_train", y_train.shape)
    torch.save(y_val,   os.path.join(fold_dir, f'val_labels_fold_{fold_number}.pt'))
    print("y_val", y_val.shape)
    torch.save(y_test,  os.path.join(fold_dir, f'test_labels_fold_{fold_number}.pt'))
    print("y_test", y_test.shape)

    logging.info(f"Guardados tensores normalizados y labels del fold {fold_number} en {fold_dir}")

    fold_number += 1


logging.info("Proceso de Cross-Validation completo.")
```
