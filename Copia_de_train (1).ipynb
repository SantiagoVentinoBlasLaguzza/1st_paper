```python
import os
import numpy as np
import matplotlib.pyplot as plt
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator

log_base_dir = '/content/drive/MyDrive/1st_paper/tensorboard_folds'
num_folds = 5

train_losses_per_fold = []
val_losses_per_fold = []

for fold_idx in range(1, num_folds+1):
    fold_log_dir = os.path.join(log_base_dir, f'fold_{fold_idx}')

    # There's typically exactly 1 tfevents file in each fold directory;
    # but if there are multiple, choose the most recent or parse them all.
    event_files = [f for f in os.listdir(fold_log_dir) if f.startswith('events.out.tfevents')]
    if not event_files:
        print(f"[WARNING] No event files found for fold {fold_idx} at {fold_log_dir}")
        train_losses_per_fold.append([])
        val_losses_per_fold.append([])
        continue

    # Sort so that we pick the latest event file if there's more than one
    event_files.sort()
    latest_event_file = os.path.join(fold_log_dir, event_files[-1])

    # Parse it
    ea = EventAccumulator(latest_event_file, size_guidance={'scalars': 0})
    ea.Reload()

    # We'll gather the 'Loss/Train' and 'Loss/Val' keys
    scalar_keys = ea.Tags()['scalars']
    train_s = []
    val_s = []

    if 'Loss/Train' not in scalar_keys or 'Loss/Val' not in scalar_keys:
        print(f"[WARNING] Missing 'Loss/Train' or 'Loss/Val' in fold {fold_idx} logs.")
        train_losses_per_fold.append([])
        val_losses_per_fold.append([])
        continue

    # Grab all train points
    for event in ea.Scalars('Loss/Train'):
        step = event.step
        value = event.value
        train_s.append((step, value))

    # Grab all val points
    for event in ea.Scalars('Loss/Val'):
        step = event.step
        value = event.value
        val_s.append((step, value))

    # Sort by step in case they are out of order
    train_s.sort(key=lambda x: x[0])
    val_s.sort(key=lambda x: x[0])

    train_losses_per_fold.append(train_s)
    val_losses_per_fold.append(val_s)

# Now let's do a combined plot
plt.figure(figsize=(8,6))
for fold_idx in range(num_folds):
    train_s = train_losses_per_fold[fold_idx]
    val_s = val_losses_per_fold[fold_idx]

    if len(train_s) == 0 or len(val_s) == 0:
        continue

    # Unzip the lists
    train_steps, train_values = zip(*train_s)
    val_steps, val_values = zip(*val_s)

    plt.plot(train_steps, train_values, label=f'Fold {fold_idx+1} Train')
    plt.plot(val_steps, val_values, '--', label=f'Fold {fold_idx+1} Val')

plt.title("Train vs. Val Loss across Folds")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()
```
