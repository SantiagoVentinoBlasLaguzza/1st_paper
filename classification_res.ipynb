```python
#!/usr/bin/env python3
"""
ensemble_classification.py

Script que:
  - Para cada fold:
    1) Carga BetaVAE entrenado de ese fold,
    2) Extrae vectores latentes mu, filtrando SOLO AD y CN (descartando "Others"),
    3) Entrena varios clasificadores + un ensamble (voting) en (train+val),
    4) Evalúa en el test del fold,
    5) Grafica y guarda métricas, AUC, etc.,
    6) (Opcional) guarda el mejor clasificador (por AUC) o el ensamble completo.
"""

import os
import sys
import pickle
import logging
import argparse
import numpy as np
import torch

from typing import Dict, List
from collections import defaultdict

# Sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import VotingClassifier

from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
    roc_curve, auc, precision_recall_curve
)

import matplotlib.pyplot as plt
import seaborn as sns

# Importa tu BetaVAE
from models.vae_res import BetaVAE

logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")


# -----------------------------------------------------
# Utilidades para graficar y evaluar
# -----------------------------------------------------
def plot_roc_curve(y_true, y_proba, classifier_name: str):
    """Grafica la curva ROC y calcula el AUC."""
    fpr, tpr, _ = roc_curve(y_true, y_proba)
    roc_auc = auc(fpr, tpr)

    plt.figure(figsize=(6, 5))
    plt.plot(fpr, tpr, color='darkorange', lw=2,
             label=f'ROC (AUC = {roc_auc:.2f})')
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.title(f'ROC - {classifier_name}')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.legend(loc="lower right")
    plt.tight_layout()
    plt.show()

def plot_pr_curve(y_true, y_proba, classifier_name: str):
    """Grafica la curva Precision-Recall."""
    precision, recall, _ = precision_recall_curve(y_true, y_proba)
    plt.figure(figsize=(6, 5))
    plt.plot(recall, precision, color='green', lw=2,
             label='PR curve')
    plt.title(f'Precision-Recall - {classifier_name}')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.legend(loc="lower left")
    plt.tight_layout()
    plt.show()

def plot_confusion_matrix(y_true, y_pred, classifier_name: str):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=["CN", "AD"], yticklabels=["CN", "AD"])
    plt.title(f'Confusion Matrix - {classifier_name}')
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.tight_layout()
    plt.show()

# -----------------------------------------------------
# Función para extraer mu solo de AD/CN
# -----------------------------------------------------
def extract_mu_ad_cn(
    vae: BetaVAE,
    data_tensor: torch.Tensor,
    labels: List[str],
    device: str = "cpu"
):
    """
    Filtra AD/CN (descarta Others).
    Devuelve X, y, con AD=1, CN=0.
    """
    vae.eval()
    X_list = []
    y_list = []
    # Asumimos data_tensor shape: (N, C, H, W) y labels de longitud N
    for i in range(data_tensor.shape[0]):
        lab = labels[i]
        # Parse label
        if '_' in lab:
            base_label, _ = lab.split('_', 1)
        else:
            base_label = lab

        if base_label not in ["AD", "CN"]:
            # skip Others
            continue

        y_val = 1 if base_label == "AD" else 0

        x = data_tensor[i].unsqueeze(0).to(device)
        with torch.no_grad():
            _, mu, _, _ = vae(x)
        X_list.append(mu.squeeze(0).cpu().numpy())
        y_list.append(y_val)

    if len(X_list) == 0:
        return None, None

    X = np.vstack(X_list)
    y = np.array(y_list, dtype=np.int32)
    return X, y


# -----------------------------------------------------
# Función principal para entrenar y evaluar clasificadores en un fold
# -----------------------------------------------------
def train_and_evaluate_fold(
    fold_idx: int,
    base_dir: str,
    latent_dim: int,
    hidden_dim: int,
    beta_val: float,
    dropout_rate: float,
    input_channels: int,
    device: str = "cpu",
    tune_all: bool = False
):
    """
    1) Carga BetaVAE de fold_idx,
    2) Extrae X,y (solo AD/CN) de train+val => data de entrenamiento,
       y test => data de test,
    3) Entrena varios clasificadores + un VotingClassifier,
    4) Devuelve métricas y registra plots.
    """
    # 1) Rutas de fold
    fold_path = os.path.join(base_dir, f"fold_{fold_idx}")
    vae_path  = os.path.join(fold_path, f"best_beta_vae_model_fold_{fold_idx}.pth")

    # 2) Cargar data
    train_data = torch.load(os.path.join(fold_path, "train_data_normed.pt"), weights_only=False)
    train_labels = torch.load(os.path.join(fold_path, f"train_labels_fold_{fold_idx}.pt"), weights_only=False)

    val_data = torch.load(os.path.join(fold_path, "val_data_normed.pt"), weights_only=False)
    val_labels = torch.load(os.path.join(fold_path, f"val_labels_fold_{fold_idx}.pt"), weights_only=False)

    test_data = torch.load(os.path.join(fold_path, "test_data_normed.pt"), weights_only=False)
    test_labels = torch.load(os.path.join(fold_path, f"test_labels_fold_{fold_idx}.pt"), weights_only=False)

    # 3) Cargar BetaVAE
    vae = BetaVAE(
        latent_dim=latent_dim,
        hidden_dim=hidden_dim,
        beta=beta_val,
        dropout_rate=dropout_rate,
        input_channels=input_channels
    ).to(device)
    vae.load_state_dict(torch.load(vae_path, map_location=device))
    vae.eval()

    # 4) Extraer latentes AD/CN para train + val
    X_train_, y_train_ = extract_mu_ad_cn(vae, train_data, train_labels, device)
    X_val_,   y_val_   = extract_mu_ad_cn(vae, val_data,   val_labels,   device)

    # Caso: si no hay AD/CN en train o val => skip
    if X_train_ is None or X_val_ is None:
        print(f"[Fold {fold_idx}] No hay sujetos AD/CN en train/val => no se entrena clasificador.")
        return None

    X_train = np.concatenate([X_train_, X_val_], axis=0)
    y_train = np.concatenate([y_train_, y_val_], axis=0)

    # 5) Extraer latentes AD/CN para test
    X_test, y_test = extract_mu_ad_cn(vae, test_data, test_labels, device)
    if X_test is None:
        print(f"[Fold {fold_idx}] No hay sujetos AD/CN en test => no se evalúa.")
        return None

    # 6) Escalado
    from sklearn.preprocessing import StandardScaler
    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train)
    X_test  = scaler.transform(X_test)

    # 7) Definir clasificadores base
    classifiers = {
        "LR": LogisticRegression(max_iter=1000, class_weight="balanced"),
        "SVM": SVC(kernel='linear', probability=True, class_weight="balanced"),
        "RF":  RandomForestClassifier(n_estimators=100, class_weight="balanced"),
        "kNN": KNeighborsClassifier(n_neighbors=5),  # k-NN no usa class_weight
        "GB":  GradientBoostingClassifier(n_estimators=100),  # no direct class_weight
        "MLP": MLPClassifier(hidden_layer_sizes=(50,), max_iter=1000),  # no direct class_weight
    }

    # Ejemplo de param grids
    param_grids = {
        "LR": {
            'C': [0.01, 0.1, 1, 10],
        },
        "SVM": {
            'C': [0.01, 0.1, 1, 10, 100],
        },
        "RF": {
            'n_estimators': [50, 100],
            'max_depth': [None, 5, 10],
        },
        "kNN": {
            'n_neighbors': [3, 5, 7],
        },
        "GB": {
            'n_estimators': [50, 100],
            'learning_rate': [0.01, 0.1],
        },
        "MLP": {
            'hidden_layer_sizes': [(50,), (100,)],
            'alpha': [1e-4, 1e-3],
        }
    }

    from sklearn.model_selection import GridSearchCV

    best_estimators = {}
    if tune_all:
        print(f"[Fold {fold_idx}] GridSearch hyperparams en TODOS los clasificadores.")
        for cname, clf in classifiers.items():
            if cname in param_grids:
                gs = GridSearchCV(
                    clf,
                    param_grid=param_grids[cname],
                    cv=3,  # mini cross-validation interno
                    scoring='roc_auc',
                    n_jobs=-1
                )
                gs.fit(X_train, y_train)
                best_estimators[cname] = gs.best_estimator_
                print(f"  => {cname} best: {gs.best_params_}, AUC={gs.best_score_:.3f}")
            else:
                best_estimators[cname] = clf
    else:
        print(f"[Fold {fold_idx}] Sin tuning. Usando clasificadores base.")
        best_estimators = classifiers

    # 8) Entrenar cada best_estimator en train
    performance = {}
    probas_dict = {}  # para curvas combinadas
    for cname, model in best_estimators.items():
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        try:
            y_prob = model.predict_proba(X_test)[:,1]
        except:
            # p.ej. SVC sin probability=True
            y_prob = None

        acc = accuracy_score(y_test, y_pred)
        prec = precision_score(y_test, y_pred, zero_division=0)
        rec = recall_score(y_test, y_pred, zero_division=0)
        f1 = f1_score(y_test, y_pred, zero_division=0)
        auc_ = roc_auc_score(y_test, y_prob) if y_prob is not None else None

        performance[cname] = {
            "Accuracy": acc,
            "Precision": prec,
            "Recall": rec,
            "F1-score": f1,
            "ROC AUC": auc_,
        }

        print(f"\n[Fold {fold_idx} - {cname}]")
        print(classification_report(y_test, y_pred, target_names=["CN","AD"]))
        plot_confusion_matrix(y_test, y_pred, classifier_name=f"Fold{fold_idx}-{cname}")

        if y_prob is not None:
            plot_roc_curve(y_test, y_prob, classifier_name=f"Fold{fold_idx}-{cname}")
            plot_pr_curve(y_test, y_prob, classifier_name=f"Fold{fold_idx}-{cname}")
            probas_dict[cname] = y_prob

    # 9) Construir un ensamble (VotingClassifier) con todos
    #    - Por defecto, 'hard' voting => usa predict()
    #    - Podemos usar 'soft' si todos tienen predict_proba (SVC con probability=True)
    #    - Ajustamos 'soft' si deseas promediar probabilidades
    clf_list = [(cn, best_estimators[cn]) for cn in best_estimators]
    voting_clf = VotingClassifier(estimators=clf_list, voting='soft')  # si SVC tiene probability=True
    voting_clf.fit(X_train, y_train)
    y_pred_ens = voting_clf.predict(X_test)
    y_prob_ens = voting_clf.predict_proba(X_test)[:,1]

    ens_acc = accuracy_score(y_test, y_pred_ens)
    ens_prec = precision_score(y_test, y_pred_ens, zero_division=0)
    ens_rec  = recall_score(y_test, y_pred_ens, zero_division=0)
    ens_f1   = f1_score(y_test, y_pred_ens, zero_division=0)
    ens_auc  = roc_auc_score(y_test, y_prob_ens)

    performance["Ensemble"] = {
        "Accuracy": ens_acc,
        "Precision": ens_prec,
        "Recall": ens_rec,
        "F1-score": ens_f1,
        "ROC AUC": ens_auc,
    }

    print(f"\n[Fold {fold_idx} - EnsembleVoting]")
    print(classification_report(y_test, y_pred_ens, target_names=["CN","AD"]))
    plot_confusion_matrix(y_test, y_pred_ens, classifier_name=f"Fold{fold_idx}-Ensemble")
    plot_roc_curve(y_test, y_prob_ens, classifier_name=f"Fold{fold_idx}-Ensemble")
    plot_pr_curve(y_test, y_prob_ens, classifier_name=f"Fold{fold_idx}-Ensemble")
    probas_dict["Ensemble"] = y_prob_ens

    # 10) Devolver performance y etc.
    return performance, probas_dict, voting_clf


def main():
    # Instead of using argparse directly, simulate command-line arguments
    # by defining a dictionary of arguments and their values.
    args_dict = {
        "base_dir": "/content/drive/MyDrive/1st_paper/",  # Replace with your actual base directory
        "num_folds": 5,
        "latent_dim": 256,
        "hidden_dim": 5120,
        "beta_val": 1.5,
        "dropout_rate": 0.0,
        "input_channels": 4,
        "tune_all": True  # Set to True if you want to tune all classifiers
    }

    # Convert the dictionary to a Namespace object, which argparse usually returns.
    class ArgsNamespace:
        def __init__(self, **kwargs):
            for k, v in kwargs.items():
                setattr(self, k, v)

    args = ArgsNamespace(**args_dict)

    device = "cuda" if torch.cuda.is_available() else "cpu"
    logging.info(f"Device = {device}")

    # Para almacenar resultados agregados
    folds_results = {}

    for fold_idx in range(1, args.num_folds+1):
        logging.info(f"=== Procesando Fold {fold_idx} ===")
        res = train_and_evaluate_fold(
            fold_idx = fold_idx,
            base_dir = args.base_dir,
            latent_dim = args.latent_dim,
            hidden_dim = args.hidden_dim,
            beta_val = args.beta_val,
            dropout_rate = args.dropout_rate,
            input_channels = args.input_channels,
            device = device,
            tune_all = args.tune_all
        )
        if res is None:
            # No hubo AD/CN, etc.
            continue
        performance, probas, ensemble_model = res
        folds_results[fold_idx] = performance

        # Guardar ensemble si deseas
        save_path = os.path.join(args.base_dir, f"fold_{fold_idx}", "best_ensemble_voting.pkl")
        with open(save_path, "wb") as f:
            pickle.dump(ensemble_model, f)
        logging.info(f"[Fold {fold_idx}] Ensemble guardado en {save_path}")

    # Al final, se pueden imprimir resumen o promedios
    # Por ejemplo, calcular el AUC medio del ensemble, etc.
    # ...

    print("\n=== Resumen de cada fold (Ensemble) ===")
    for f_idx in folds_results:
        perf = folds_results[f_idx]
        if "Ensemble" in perf:
            e = perf["Ensemble"]
            print(f"Fold {f_idx} => Ensemble AUC={e['ROC AUC']:.3f}, Acc={e['Accuracy']:.3f}")

    print("Fin.")


if __name__ == "__main__":
    main()
```
